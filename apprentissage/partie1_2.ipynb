# Dataframe
Module Pandas. Fonction de lecture de fichiers CSV (read_csv), découper contenu et le stocker dans un tableau. 

```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('future.no_silent_downcasting', True)
```

# Libellés des colonnes

```python
nosPokemons = pd.read_csv("datas/pokedex.csv")

#Affichage des colonnes du Dataframe
print(nosPokemons.columns.values)
```

# Affichage des 10 premières lignes de données

```python
print(nosPokemons.head(10))
```

# Transformer en feature utilisable : vrai = 1, faux = 0

```python
# remplacement de VRAI FAUX par 1 0
nosPokemonsClean = nosPokemons
nosPokemonsClean["LEGENDAIRE"] = nosPokemonsClean["LEGENDAIRE"].replace("VRAI", 1)
nosPokemonsClean["LEGENDAIRE"] = nosPokemonsClean["LEGENDAIRE"].replace("FAUX", 0)
nosPokemonsClean["LEGENDAIRE"] = nosPokemonsClean["LEGENDAIRE"].astype(int)
print(nosPokemonsClean.head())
```

# Fonction shape() du Dataframe : données manquantes?
Pour voir s'il manque des données
Significations du résultat : (observations, caractéristiques)

```python
#Comptage du nombre d'observations et de features
print(nosPokemonsClean.shape)
```

# Fonction info() du Dataframe : Nombre d'éléments par feature
Pour voir s'il y a des observations manquantes.

Résultat : 
- manque 1 nom, car 799 éléments
- manque 387 Type_2, car 413

```python
#Informations sur notre jeu de données
print(nosPokemonsClean.info())
```

Rechercher et afficher l'observation du Dataframe nosPokemon. Nom is null pour trouver le feature manquant.
